# Renewable Energy Capacity — 2010 to 2023

This notebook tracks the growth of solar and wind power installations
worldwide using data from the International Renewable Energy Agency (IRENA).

## Scope
- **Technologies**: Utility-scale solar PV, onshore wind, offshore wind
- **Geography**: Global totals, with breakdowns for EU, US, China, and India
- **Unit**: Gigawatts (GW) of installed capacity

```python
solar_2010 = 40
solar_2023 = 1600
print(f'Global solar PV capacity 2010: {solar_2010} GW')
print(f'Global solar PV capacity 2023: {solar_2023} GW')
print(f'Growth factor: {solar_2023 // solar_2010}×')
```

## Key Trends

- **Solar PV** capacity grew **40-fold** between 2010 and 2023,
  driven by cost reductions of over 90 %.
- **Wind** (onshore + offshore) nearly quadrupled from ~200 GW to ~950 GW.
- **China** leads with ~500 GW of solar and ~400 GW of wind installed.
- The **levelised cost of electricity (LCOE)** for utility solar is now
  below $0.03 / kWh in many regions, cheaper than any fossil fuel source.

## Policy Drivers

Subsidies, feed-in tariffs, and renewable portfolio standards have been
the primary policy mechanisms, with the US Inflation Reduction Act (2022)
injecting $369 billion into clean energy incentives.

## Conclusion

Renewable energy deployment is accelerating rapidly. If current growth
rates continue, solar and wind alone could supply over 50 % of global
electricity by 2035, making deep decarbonisation of the power sector
technically and economically feasible.
