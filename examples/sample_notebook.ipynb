# Climate Data Analysis

This notebook explores global temperature trends from 1880 to 2023
using the NASA GISS Surface Temperature dataset.

## Dataset
- **Source**: NASA GISS (Goddard Institute for Space Studies)
- **Records**: 143 years of monthly global temperature anomalies
- **Baseline**: 1951–1980 average

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('nasa_giss.csv')
print(f'Dataset loaded: {df.shape[0]} rows × {df.shape[1]} columns')
```

## Key Findings

1. **Warming trend**: The global mean temperature anomaly has risen by
   approximately **+1.2 °C** since pre-industrial times.
2. **Acceleration**: The rate of warming roughly doubled after 1975
   compared to the preceding century.
3. **Record years**: 2016 and 2023 are the two hottest years on record,
   driven by a combination of long-term warming and El Niño events.

```python
early = df[df['year'] <= 1940]['anomaly'].mean()
recent = df[df['year'] >= 1980]['anomaly'].mean()
print(f'Mean anomaly 1880-1940: {early:+.2f}°C')
print(f'Mean anomaly 1980-2023: {recent:+.2f}°C')
print(f'Difference: {recent - early:+.2f}°C')
```

## Conclusion

The data clearly shows an upward trend in global surface temperatures
consistent with anthropogenic greenhouse gas emissions. Mitigation
strategies should target a maximum of +1.5 °C above pre-industrial
levels as agreed in the Paris Accord.
